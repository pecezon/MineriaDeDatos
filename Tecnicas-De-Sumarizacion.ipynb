```python
import pandas as pd

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

```python
# Skewness
asimetria_edad = df['Age'].skew()
print(f"Asimetría (Skewness) de Age: {asimetria_edad}")
```

El resultado nos da 0.38 un sesgo hacia la derecha, significa qu ela mayoria de los pasajeros estaba concentrada en edades bajas, habia pocos ancianos

```python
# Curtosis
curtosis_tarifa = df['Fare'].kurt()
print(f"Curtosis de Fare: {curtosis_tarifa}")
```

El valor es alto y positivo, significa que la distribucion tiene colas muy pesadas osea outliers grandes y varios, hay precios de boletos extremos

```python
# Cross tabulation
tabla_cruzada = pd.crosstab(df['Pclass'], df['Survived'], margins=True, margins_name="Total")
display(tabla_cruzada)
```

El patron al hacer la tabla es la desigualdad socioeconomica entre mas rico mas habia probabilidad de tu salvacion, la primera clase tiene mas sobrevivientes a fallecidos, en la 3era clase sucede todo lo contrario
