```python
import random 
random_list = [random.random() for _ in range(50_000_000)]
print ("Random list length:", len(random_list))
```

```python

def calculate_average(random_list):
    avg = 0
    for i in random_list:
        avg += i
    avg /= len(random_list)
    print("Average:", avg)
    return avg

def calculate_std(random_list, avg):
    std = 0
    for i in random_list:
        std += (i - avg) ** 2
    std /= len(random_list)
    std = std ** 0.5
    print("Standard deviation:", std)
    return std

avg = calculate_average(random_list)
std = calculate_std(random_list, avg)


```
