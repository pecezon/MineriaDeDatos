```python
import random 
random_list = [random.random() for _ in range(50_000_000)]
print ("Random list length:", len(random_list))
```

```python

def calculate_average(random_list):
    avg = 0
    for i in random_list:
        avg += i
    avg /= len(random_list)
    print("Average:", avg)
    return avg

def calculate_std(random_list, avg):
    std = 0
    for i in random_list:
        std += (i - avg) ** 2
    std /= len(random_list)
    std = std ** 0.5
    print("Standard deviation:", std)
    return std

avg = calculate_average(random_list)
std = calculate_std(random_list, avg)


```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

#print(df.head())
print(df.describe())
```

```python
women = 0
men = 0
total_women = 0
total_men = 0

# Iterate for each row in the DataFrame
for row in df.iterrows():
    if row[1]['Sex'] == 'female':
        women += 1
        total_women += row[1]['Fare']
    else:
        men += 1
        total_men += row[1]['Fare']

print("Total women:", women)
print("Total men:", men)
print("Average fare for women:", total_women / women)
print("Average fare for men:", total_men / men)    

# Using group by
df.groupby('Sex')['Fare'].mean()

df['Sex'].value_counts()

```

```python
fare_IQR = df["Fare"].quantile(0.75) - df["Fare"].quantile(0.25)

outlier_count = 0

for row in df.iterrows():
    if row[1]['Fare'] < (df["Fare"].quantile(0.25) - 1.5 * fare_IQR) or row[1]['Fare'] > (df["Fare"].quantile(0.75) + 1.5 * fare_IQR):
        outlier_count += 1

print("Number of outliers in Fare:", outlier_count)
```

```python
import matplotlib.pyplot as plt

plt.show()

#Box Plot
df.boxplot(column='Fare', by='Sex')
```
