```python
import pandas as pd
from datetime import date
train_df = pd.read_csv("train_p2.csv")
test_df = pd.read_csv("test_p2.csv")
```

```python
display(train_df.head())
display(test_df.head())
```

```python
# Pista 1 y 2 verificacion encoding (label)
mapeo_verificacion = {
    'NINGUNO': 0,
    'TELEFONO': 1,
    'CORREO': 1,
    'VERIFICADO': 2,
}

train_df['verfificacion_encoded'] = train_df['nivel_verificacion'].map(mapeo_verificacion)

print("Resultados del Label Encoding\n")
display(train_df[['user_id', 'nivel_verificacion', 'verfificacion_encoded']].head())

train_df = train_df.drop(columns=["nivel_verificacion"])

train_df = pd.get_dummies(train_df, columns=['pais', 'dispositivo', 'email_domain'], dtype=int)

print("Resultados del One-Hot Encoding\n")
pd.set_option('display.max_columns', None)
display(train_df.head())

#Pista 1 verificacion encoding (label)

test_df['verfificacion_encoded'] = test_df['nivel_verificacion'].map(mapeo_verificacion)

print("Resultados del Label Encoding\n")
display(test_df[['user_id', 'nivel_verificacion', 'verfificacion_encoded']].head())

test_df = test_df.drop(columns=["nivel_verificacion"])

test_df = pd.get_dummies(test_df, columns=['pais', 'dispositivo', 'email_domain'], dtype=int)

print("Resultados del One-Hot Encoding\n")
pd.set_option('display.max_columns', None)
display(test_df.head())
```

```python
import numpy as np

# Paso 3
# Para hora pico
# Train
train_df['hora_pico_actividad'] = train_df['hora_pico_actividad'].astype(int)

train_df["hour_sin"] = train_df['hora_pico_actividad'].apply(lambda x: np.sin(x / 24 * 2 * np.pi))
train_df["hour_cos"] = train_df['hora_pico_actividad'].apply(lambda x: np.cos(x / 24 * 2 * np.pi))
    
train_df.drop(columns=["hora_pico_actividad"])

train_df["month"] = train_df['fecha_creacion'].apply(lambda x: int(x.split("-")[1]))
train_df["day"] = train_df['fecha_creacion'].apply(lambda x: int(x.split("-")[2]))

train_df["month_sin"] = train_df['month'].apply(lambda x: np.sin(x / 12 * 2 * np.pi))
train_df["month_cos"] = train_df['month'].apply(lambda x: np.cos(x / 12 * 2 * np.pi))

train_df["day_sin"] = train_df['day'].apply(lambda x: np.sin(x / 31 * 2 * np.pi))   
train_df["day_cos"] = train_df['day'].apply(lambda x: np.cos(x / 31 * 2 * np.pi))

train_df.drop(columns=["month", "day"])

# Test
test_df['hora_pico_actividad'] = test_df['hora_pico_actividad'].astype(int)
test_df["hour_sin"] = test_df['hora_pico_actividad'].apply(lambda x: np.sin(x / 24 * 2 * np.pi))
test_df["hour_cos"] = test_df['hora_pico_actividad'].apply(lambda x: np.cos(x / 24 * 2 * np.pi))
test_df.drop(columns=["hora_pico_actividad"])

test_df["month"] = test_df['fecha_creacion'].apply(lambda x: int(x.split("-")[1]))
test_df["day"] = test_df['fecha_creacion'].apply(lambda x: int(x.split("-")[2]))

test_df["month_sin"] = test_df['month'].apply(lambda x: np.sin(x / 12 * 2 * np.pi))
test_df["month_cos"] = test_df['month'].apply(lambda x: np.cos(x / 12 * 2 * np.pi))

test_df["day_sin"] = test_df['day'].apply(lambda x: np.sin(x / 31 * 2 * np.pi))   
test_df["day_cos"] = test_df['day'].apply(lambda x: np.cos(x / 31 * 2 * np.pi))

train_df.drop(columns=["month", "day"])
```

```python
# Pista 4
train_df['seguidores_log1p'] = np.log1p(train_df['seguidores'])
train_df['siguiendo_log1p'] = np.log1p(train_df['siguiendo'])
train_df['posts_log1p'] = np.log1p(train_df['posts_totales'])
train_df['likes_log1p'] = np.log1p(train_df['likes_given_30d'])
# train_df = train_df.drop(columns=["seguidores", "siguiendo", "posts_totales", "likes_given_30d"])
train_df.head()

test_df['seguidores_log1p'] = np.log1p(test_df['seguidores'])
test_df['siguiendo_log1p'] = np.log1p(test_df['siguiendo'])
test_df['posts_log1p'] = np.log1p(test_df['posts_totales'])
test_df['likes_log1p'] = np.log1p(test_df['likes_given_30d'])
# test_df = test_df.drop(columns=["seguidores", "siguiendo", "posts_totales", "likes_given_30d"])
test_df.head()
```

```python
# Pista 5

# Post por dia promedio = post creados / (hoy - Dia de creacion)
# Train
train_df["tiempo_creado"] = (
    pd.Timestamp(today) - pd.to_datetime(train_df["fecha_creacion"])
).dt.days
# Test
test_df["tiempo_creado"] = (
    pd.Timestamp(today) - pd.to_datetime(test_df["fecha_creacion"])
).dt.days

# Train
train_df["avg_post_per_day"] = train_df["posts_totales"] / train_df["tiempo_creado"]
# Test
test_df["avg_post_per_day"] = test_df["posts_totales"] / test_df["tiempo_creado"]

# Seguidores / Siguiendo 
# Train
train_df["seguidor_seguidores_ratio"] = train_df["seguidores"] / train_df["siguiendo"]
# Test
test_df["seguidor_seguidores_ratio"] = test_df["seguidores"] / test_df["siguiendo"]

#  likes_given_30d /  avg_session_min 
# tasa_interaccion / posts_totales
# Train
train_df["avg_sessiontime_likes"] = (train_df["likes_given_30d"]/30) / (train_df["avg_session_min"])
# Test
test_df["avg_sessiontime_likes"] = (test_df["likes_given_30d"]/30) / (test_df["avg_session_min"])


display(train_df.head())
```

```python
# Pista 6
from sklearn.cluster import KMeans

# Entrenar con train
df_kmeans = train_df[['edad_usuario', 'variance_post_interval_hr']].copy()
kmeans = KMeans(n_clusters=2, random_state=0, n_init="auto").fit(df_kmeans)
train_df["kmeans_age_variance_hr"] = kmeans.labels_

# Test
test_df["kmeans_age_variance_hr"] = kmeans.predict(test_df[['edad_usuario', 'variance_post_interval_hr']])

display(train_df.head())
```
