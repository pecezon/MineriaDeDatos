## Analisis de Vuelos

```python
import pandas as pd

df = pd.read_csv('vuelos.csv')
display(df.head(5))

len(df)
```

### Filtrar

```python
# Eliminar valores nulos
df = df.dropna()

# All caps
df['Origen'] = df['Origen'].str.upper()
df['Destino'] = df['Destino'].str.upper()

# Drop same origen y destino
df = df[df['Origen'] != df['Destino']]

# Quitar distancias negativas
df = df[df['Distancia_km'] >= 0]

# Quitar rutas duplicadas
df = df.drop_duplicates(subset=['Origen', 'Destino'])

len(df)

```

### Crear Grafo

```python
import networkx as nx

grafo = nx.from_pandas_edgelist(df,
    source='Origen',
    target='Destino',
    edge_attr=None,
    create_using=None,
    edge_key=None,)

print('Vertices:', list(grafo.nodes()))
print('Aristas:', list(grafo.edges(data=True)))

```

### Matriz


```python
matriz = nx.adjacency_matrix(grafo).todense()
print("Memoria Matriz: ", matriz.nbytes / (1024 * 1024), "MB")
print("Memoria DF: ", df.memory_usage(deep=True).sum() / (1024 * 1024), "MB")

```

**¿Cuántas veces más pesada es la Matriz en comparación con la Lista de Aristas y por qué sucede esto?** 80 veces mas pesada, por que la matriz guarda un monton de 0's con un tamano NxN, el DF solo guarda las aristas la matriz todas las combinaciones de datos posibles aunque no haya conexion
